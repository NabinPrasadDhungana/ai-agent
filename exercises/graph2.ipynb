```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    operation: str
    result: str | None
```

```python
def operate(state: AgentState) -> AgentState:
    operation = state["operation"]
    if state["operation"] not in ["*", "+"]:
        state["result"] = "Invalid input for operation, only '*' and '+' are allowed."
        return state
        
    if operation == '*':
        result = 1
        for i in state["values"]:
            result = i * result
        state["result"] = f"Hi {state["name"]}, you answer is: {result}."
        
        return state
        
    result = 0
    for i in state["values"]:
        result = result + i
    state["result"] = f"Hi {state["name"]}, you answer is: {result}."
    
    return state
```

```python
graph_builder = StateGraph(AgentState)

graph_builder.add_node("operate", operate)

graph_builder.set_entry_point("operate")
graph_builder.set_finish_point("operate")

app = graph_builder.compile()
```

```python
app.invoke({
        'values': [1, 2, 3, 4],
        'name': "Nabin",
        'operation': "*"
        }
)
```
