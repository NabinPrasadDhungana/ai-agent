```python
from typing import TypedDict, List
import random
from langgraph.graph import StateGraph, START, END
```

```python
class AgentState(TypedDict):
    name: str
    num: List[int]
    counter: int
```

```python
def greeting(state: AgentState) -> AgentState:
    """This function is used to greet a person."""
    state['counter'] = 0
    state['name'] = f"Hello {state['name']}"
    
    return state
```

```python
def random_node(state: AgentState) -> AgentState:
    """This function generates random integers and updates the state."""  
    number = random.randint(1, 5)
    state['num'].append(number)
    state['counter'] += 1
    
    return state
```

```python
def conditional_node(state: AgentState):
    """This function conditionally routes back to the random_node node."""            
    if state['counter'] <= 4:
        print(f"Enterring Loop {state['counter']}")
        return "loop"
    
    return "exit"
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeting", greeting)
graph.add_node("random", random_node)

graph.add_edge(START, "greeting")
graph.add_edge("greeting", "random")


graph.add_conditional_edges(
    "random",
    conditional_node,
    
    {
        "loop": "random",
        "exit": END
        }
)


app = graph.compile()
```

```python
app.invoke({
    "name": "Nabin",
    "num": [],
    "counter": 7,
})
```
